# 1. Download structural ensemble from RCSB PDB 

## What is the goal of this notebook?


You can use this notebook to download all the structures that match a specific 'keyword' and that are deposited on the PDB. For example, in this tutorial, we will download all the stuctures that contain the keyword 'triosephosphate isomerase' in their metadata. 

You can also download structures based on the sequence of the molecule of interest. 

You can choose to use only one of the methods, or you can do both if you want to make sure that you obtain all possible matching structures deposited in the PDB. 

Optional: You can also download the [computed structure models](https://www.rcsb.org/docs/general-help/computed-structure-models-and-rcsborg) (CSMs), predicted by AlphaFold, that the RCSB PDB also offers.

*NOTE: sequences and structures are downloaded from the [wwPDB beta archive](https://www.wwpdb.org/ftp/pdb-beta-ftp-sites), which will replace the current archive on July 21, 2027. It uses the extended PDB IDs: the legacy ID `1aw1` becomes `pdb_00001aw1`, and the files are named after it (e.g. `pdb_00001aw1+01.cif`).*



## Import libraries, including CIFcurator

```python
import numpy as np
from CIFcurator import cifcurator_io, pdbutils
```

## Define and create working directory (where you want to save your structures)

```python
PROJDIR="./TIM/"
```

With check_project we can check if our project directory (PROJDIR) exists, and if it doesn't, we create it. 

```python
cifcurator_io.check_project(projdir=PROJDIR)
```

## Check if the file with all sequences from PDB already exists. 

'seqfile' is where we will download the sequences of all the structures currently deposited in the PDB (`pdb_seqres.fasta.gz` from the [wwPDB beta archive](https://www.wwpdb.org/ftp/pdb-beta-ftp-sites), decompressed).
If you haven't downloaded them, or if you want to update the list, keep:
> update=True

otherwise change to:
>update=False 

And make sure to change 'seqfile' to the path where you have stored the sequences. 


```python
update=True            
seqfile=PROJDIR+'seqres.txt' 
```

## Retrieve reference sequence(s) from keyword

Change the keyword to any protein (or any other biomolecule) of interest. 


```python
keyword='triosephosphate isomerase' 
```

Run the next cell. 

The next cell will download all the sequences currently deposited in the PDB, from the wwPDB beta archive.

It will use the given 'keyword' to search for all the matching sequences, based on the metadata (mode='metadata'). \
Results will be stored in 'ref_sequences' and 'ref_metadata', and printed to screen. 

Note: Depending on your internet connection, it may take a few minutes to download the sequences. But you only need \
to do this step once. 

```python
ref_sequences, ref_metadata = pdbutils.retrieve_sequence_from_PDB(keyword, mode='metadata', update=True, seqfile=seqfile)

print('{0} sequences were identified as potential hits! \n'.format(len(ref_sequences)))

for iseq in np.arange(len(ref_sequences)):
    print('{0} {1}'.format(ref_metadata[iseq], ref_sequences[iseq]))
```

## Retrieve all sequences that match the reference sequence(s)

We will use the sequences we retrieved in our previous search (mode='metadata') to find all the chains in the PDB whose sequence matches one of them: a chain matches when its sequence ends with one of the reference sequences.

`retrieve_sequences_matching_from_PDB` reads the PDB sequences we downloaded in the previous step only once, for all the reference sequences, and lists each chain once. (Earlier versions of this notebook called `retrieve_sequence_from_PDB(seq, mode='sequence')` once per reference sequence, which found the same chains but read the whole file every time: about 30 minutes for triosephosphate isomerase, instead of a few seconds.)

We will print to screen all the sequences that are retrieved. 

```python
sequences, metadata = pdbutils.retrieve_sequences_matching_from_PDB(ref_sequences, seqfile)

print('{0} sequences were retrieved! \n'.format(len(sequences)))

for iseq in np.arange(len(sequences)):
    print('{0} {1}'.format(metadata[iseq], sequences[iseq]))
```

## Optional: download computed structure models (CSMs)

Besides experimental structures, RCSB PDB also offers [computed structure models](https://www.rcsb.org/docs/general-help/computed-structure-models-and-rcsborg) (CSMs), predicted by AlphaFold. **This step is optional**: skip it if you only want experimental structures. To include CSMs, set `download_csm=True` in the next cell.

`download_csm_from_keyword` finds the CSMs that match the keyword with the RCSB full text search (the same search as the RCSB PDB web site, so it can find a different set of proteins than the search above), and downloads their mmCIF files from [AlphaFold DB](https://alphafold.ebi.ac.uk) into the same 'raw_bank' as the experimental structures. Each model is saved as `CSMID+01.cif`, where CSMID is the RCSB ID of the model (e.g. `AF_AFP60174F1+01.cif`). Models from other databases (ModelArchive, IDs starting with `MA_`) are skipped.

A search can find many models (more than 700 for triosephosphate isomerase), so `max_models` limits how many are downloaded, best matches first. In this tutorial we download only 10 models (`max_models=10`). **To download all the models that match the keyword, set `max_models=None`.**

**Before including CSMs, keep in mind:**
* A CSM contains **one protein chain only**: no nucleic acids, ligands, waters or other copies of the protein. For example, a Cas9 model has no guide RNA or target DNA, and a triosephosphate isomerase model is a monomer, while the experimental structures are dimers.
* A CSM covers the full UniProt sequence, including parts that are usually missing from experimental structures, such as signal peptides.
* The B-factor column of a CSM holds the model confidence (pLDDT), not a B-factor.
* The entity name of a CSM comes from UniProt, and can differ from the names in the experimental structures (e.g. `TRIOSEPHOSPHATE ISOMERASE, CYTOSOLIC`). In Step 3 each name is a separate entity that needs its own chain IDs.

```python
download_csm=True  # set to False if you do not want to download computed structure models
max_models=10        # 10 models for this tutorial. Set max_models=None to download ALL the models

if download_csm:
    csmids = pdbutils.download_csm_from_keyword(keyword, projdir=PROJDIR, max_models=max_models)
```
